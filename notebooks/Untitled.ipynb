```python
import requests



BASE_URL = "https://api.artic.edu/api/v1"

def get_artworks(limit=10, page=1):
    url = f"{BASE_URL}/artworks/search"

    params = {
        "q": "This viewpoint",
        "limit": 10,
        "fields": "id,title,alt_text,description,short_description,style_title,theme_titles,subject_titles,place_of_origin",
        "page" : page
    }

    response = requests.get(url, params=params)
    response.raise_for_status()

    return response.json()

data = get_artworks(5, 2)
#print(data)

#print(data.keys()) # keys are 'pagination', 'data', 'info' and 'config'
#print(len(data["data"]))

artwork = data["data"][0]
print(artwork)
print(artwork["id"])
print(artwork["title"])

# usefuls fields : title, alt_text, description, short_description, style_title, theme_title
```

```python
import requests


BASE_URL = "https://api.artic.edu/api/v1/artworks/search"


def search_artworks(query, limit=10):
    params = {
        "q": query,
        "limit": limit,
        "fields": (
            "id,"
            "title,"
            "alt_text,"
            "description,"
            "short_description,"
            "subject_titles,"
            "theme_titles,"
            "style_titles,"
            "image_id,"
            "is_public_domain"
        )
    }

    response = requests.get(BASE_URL, params=params)
    response.raise_for_status()

    return response.json()
```

```python
results = search_artworks("Paris")

for artwork in results["data"]:
    print(artwork["id"], artwork["title"])
    
```
